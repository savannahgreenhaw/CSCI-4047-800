### 1. A tiny, dirty dataset of quiz scores (out of 100) and letter grades is written below

```python
import pandas as pd
import numpy as np

# A tiny, dirty dataset of quiz scores (out of 100) and letter grades
data = {
    'name':  ['Amir', 'Bella', 'Chen', 'Diego', 'Eve', 'Farah'],
    'score': [88, None, 95, 72, 200, 68],
    'grade': ['A', 'B', 'A', 'c', 'B', 'b'],
}
df = pd.DataFrame(data)
print("Before:")
print(df)
```

### 2. Preprocess
- first step below will be done for you

```python
# 1. NOISE: a score of 200 is impossible (max is 100). Mark it invalid.
df.loc[df['score'] > 100, 'score'] = np.nan

# 2. MISSING: fill the empty scores with the column mean (imputation)
    # hint: use the pandas fillna function to replace the None and and 200 with a mean value.
df['score'] = df['score'].fillna(df['score'].mean())

# 3. INCONSISTENT: the grades mix upper and lower case. Standardize them.
df['grade'] = df['grade'].str.upper()
# 4. Print the df again below
print(df)
# 5. Why mark it as missing first, instead of replacing it with the mean directly?
# First we havr to replace the 200 because it is invalid and we dont want it to be inculded when calculating the mean.
```
